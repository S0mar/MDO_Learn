# Problem 1.2

Plotting a two-dimensional function. <br>
Consider the two-dimensional function: <br>
f(x1, x2) = x1^3 + 2*x1*x2^2 - x2^3 - 20*x1 <br>
Plot the function contours and find the approximate location of
the minimum point(s). Is there a global minimum? <br>
Exploration: <br>
Plot other functions to get an intuition about their trends and
minima. You can start with simple low-order polynomials and
then add higher-order terms, trying different coefficients. Then
you can also try nonalgebraic functions. This will give you an
intuition about the function trends and minima.

```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize
```

## Approach

```python
# define function to explore
def func(x):
    return x[0]**3 + 2*x[0]*x[1]**2 - x[1]**3 - 20*x[0]

#plt.style.use('_mpl-gallery-nogrid')
#plt.style.use('_mpl-gallery')

# make data
X1, X2 = np.meshgrid(np.linspace(-5, 5, 500), np.linspace(-5, 5, 500))
Z = func([X1, X2])
levels = np.linspace(np.min(Z), np.max(Z), 21)

# plot
fig1, ax1 = plt.subplots()
ax1.contourf(X1, X2, Z, levels=levels)
cbar = fig1.colorbar(ax1.collections[0], ax=ax1)
ax1.plot(-2.58,0,'o', color='red') #local maximum
ax1.plot( 2.58,0,'o', color='red') #local minimum
ax1.plot(1.75,2.33,'o', color='yellow') # saddle point
ax1.plot(-1.75,-2.33,'o', color='yellow') # saddle point

cbar.set_label("f(x1, x2)")

fig2, ax2 = plt.subplots(subplot_kw={"projection": "3d"})
ax2.plot_surface(X1, X2, Z, vmin=Z.min(), cmap="Blues")

plt.show()
```

```python
# define function to explore
def func2(x):
    return 0.5*np.sin(x[0]) * 2*np.cos(x[1])

#plt.style.use('_mpl-gallery-nogrid')
#plt.style.use('_mpl-gallery')

# make data
X1, X2 = np.meshgrid(np.linspace(-10, 10, 500), np.linspace(-10, 10, 500))
Z = func2([X1, X2])
levels = np.linspace(np.min(Z), np.max(Z), 21)

# plot
fig1, ax1 = plt.subplots()
ax1.contourf(X1, X2, Z, levels=levels)

fig2, ax2 = plt.subplots(subplot_kw={"projection": "3d"})
ax2.plot_surface(X1, X2, Z, vmin=Z.min(), cmap="Blues")

plt.show()
```

## Results

### Exploration of func: <br>
Global minimum: no global minimum. for x1 -> -inf and x2 -> inf the function goes to -inf. <br>
Points of interes: visual approximation show that there are four local point of interest. one local maximum at [-2.58, 0], a local minimum at [2.58, 0] and two saddle points at [+-1.75,+-2.33]

## Interpretation

local minima or other points of interes are hard to find just by looking at it. calculating them is the better option. even for simple funcitons like this
