```python
g_acc = 9.81

def travel_time(y_int, n, mu_k=0.0, h=1.0, x_end=1.0, y_end=0.0):
    """Discretised brachistochrone objective (D.1.7), without the constant sqrt(2/g).
    y_int: the n-2 interior heights; nodes uniformly spaced in x from (0, h) to (x_end, y_end)."""
    x = np.linspace(0.0, x_end, n)
    y = np.concatenate(([h], np.asarray(y_int, dtype=float), [y_end]))
    dx, dy = np.diff(x), np.diff(y)
    root = np.sqrt(h - y - mu_k * x)
    return np.sum(np.sqrt(dx ** 2 + dy ** 2) / (root[1:] + root[:-1]))

def cycloid(theta, a=0.572917):
    """Exact frictionless solution, theta in [0, 2.412]."""
    return a * (theta - np.sin(theta)), 1 - a * (1 - np.cos(theta))

# multiply travel_time by np.sqrt(2 / g_acc) to get seconds
mu_k = 0.3
```
