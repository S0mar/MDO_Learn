# Problem 4.4

Review Kepler's wine barrel story from Section 2.2. Approximate the barrel as a cylinder and find the height and diameter of a barrel that maximizes its volume for a diagonal measurement of 1 m.

```python
%load_ext autoreload
%autoreload 2
import numpy as np
import matplotlib.pyplot as plt
from scipy import optimize
```

## Approach

No code scaffold: this is an analytical problem (use code only to check).

$h$ = height, $r$ = radius, $a$ = diagonal, $v$ = volume (all values have to be positive)

$$a^2 = {\left(\frac{h}{2}\right)^2 + (2r)^2}$$

$$v = \pi r^2 h$$

Maximize $v$.

Rearrange $a$ for $r$:

$$r^2 = \frac{a^2}{4} - \frac{h^2}{16}$$

Substitute $r$ into $v$:

$$v = \pi \left(\frac{a^2}{4} - \frac{h^2}{16}\right) h = \frac{\pi h a^2}{4} - \frac{\pi h^3}{16}$$

Find the maximum volume:

$$\frac{dv}{dh} = 0 = \frac{\pi a^2}{4} - \frac{3\pi h^2}{16}$$

Rearrange for $h$:

$$h = \sqrt{\frac{4}{3}\, a^2} = \frac{2a}{\sqrt{3}}$$

With $a = 1$ (and considering that $h$ has to be positive):

$$h = \frac{2}{\sqrt{3}} = 1.1547$$

Put $h$ into $r$ to get the final dimensions:

$$r^2 = \frac{1^2}{4} - \frac{\left(2/\sqrt{3}\right)^2}{16} = \frac{1}{4} - \frac{4/3}{16} = \frac{1}{4} - \frac{1}{12} = \frac{1}{6}$$

$$r = 0.4082$$

Final dimensions: $a = 1$, $h = 1.1547$, $r = 0.4082$ (diameter $d = 2r = 0.8165$)

Check that this is a maximum:

$$\frac{d^2 v}{dh^2} = -\frac{3\pi h}{8} < 0 \quad \text{for } h > 0$$

Calculate the final volume:

$$v_{max} = \pi r^2 h = \pi \cdot 0.4082^2 \cdot 1.1547 = 0.6046\ \text{m}^3$$

```python
# validate numbers with code

A_TARGET = 1.0   # diagonal [m]

# maximize v  ->  scipy minimizes, so minimize -v

def volume(x):
    r, h = x
    return np.pi * r**2 * h

def objective(x):
    return -volume(x)  # negative sign because we want to maximize volume

def diagonal(x):
    r, h = x
    return np.sqrt((h/2)**2 + (2*r)**2)

def diagonal_constraint(x):
    return diagonal(x) - A_TARGET  # equals 0 when feasible

# starting values and bounds
x0 = np.array([0.5, 1.0])  # [r, h]
bounds = [(0, 2), (0, 2)]  # r > 0, h > 0

# solve
res = optimize.minimize(objective, x0, method="SLSQP", bounds=bounds,
                         constraints=[{"type": "eq", "fun": diagonal_constraint}])

# compare with analytical result
r_opt, h_opt = res.x
print(res.success, res.message)
print(f"r = {r_opt:.4f} (analytical 0.4082)")
print(f"h = {h_opt:.4f} (analytical 1.1547)")
print(f"v = {volume(res.x):.4f} (analytical 0.6046)")
print(f"diagonal = {diagonal(res.x):.6f} (should be 1)")
```

## Results

the analytical values match the code values. the resulting maximum volume is around 0.6m^3. h/d = sqrt(2) which is the result kepler found. the optimum is quite flat so small changes in dimensions dont change the Volume by much (a change in 10% in h only changes the volume by ~1.5%).
