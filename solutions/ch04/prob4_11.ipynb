## Approach

```python
# Wing design problem, Appendix D.1.6: design variables b (span) and c (chord), rectangular wing, S = b c
par = dict(
    rho=1.2,        # air density [kg/m^3]
    mu=1.8e-5,      # air viscosity [kg/(m s)]
    k=1.2,          # form factor
    CL=0.4,         # lift coefficient
    e=0.80,         # Oswald efficiency factor
    W0=1000.0,      # fixed weight [N]
    Ws=8.0,         # wing-area-dependent weight [N/m^2]
    eta_max=0.8,    # peak propulsive efficiency
    v_bar=20.0,     # speed at peak efficiency [m/s]
    sigma=5.0,      # width of the efficiency curve [m/s]
)

def power(x, p=par):
    """Power required for level flight, P = D v / eta, for x = [b, c]."""
    b, c = x
    S = b * c
    W = p["W0"] + p["Ws"] * S                      # lift = weight
    v = np.sqrt(2 * W / (p["rho"] * p["CL"] * S))  # speed that gives CL at this weight
    q = 0.5 * p["rho"] * v ** 2
    Re = p["rho"] * v * c / p["mu"]
    Cf = 0.074 / Re ** 0.2
    D_f = p["k"] * Cf * q * 2.05 * S
    D_i = W ** 2 / (q * np.pi * b ** 2 * p["e"])
    eta = p["eta_max"] * np.exp(-((v - p["v_bar"]) ** 2) / (2 * p["sigma"] ** 2))
    return (D_i + D_f) * v / eta

# reference from the book: b = 25.48 m, c = 0.50 m
```
